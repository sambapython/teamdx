```python
from faker import Faker
from random import randint
fake=Faker()
con = psycopg2.connect(host="localhost", user="postgres", password="postgres", port=5432,dbname="db1")
cur = con.cursor()
for i in range(3,1000):
    query=f"insert into person values({i},'{fake.name()[:19]}',{randint(23,60)},{randint(1000000,3000000)})"
    cur.execute(query)
con.commit()
```

```python
cur.execute("select * from person")
```

```python
data = cur.fetchall()
print(data)
```

```python
sum_sal = 0
for row in data:
    sal=row[-1]
    sum_sal = sum_sal+sal
print(sum_sal)
```
